## 1. Environment & Setup
We begin by importing our core libraries.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print(f"Numpy:  {np.__version__}")
print(f"Pandas: {pd.__version__}")
```

## 2. Dataset Generation
Generating simulated user traffic and conversion rates:

```python
dates = pd.date_range("2026-01-01", periods=5)
df = pd.DataFrame({
    "Date": dates,
    "Organic_Users": [1240, 1420, 1380, 1650, 1890],
    "Ad_Clicks": [450, 520, 610, 580, 720],
    "Conversion_Rate": [0.038, 0.042, 0.045, 0.041, 0.049]
})
df
```

## 3. Harmonic Signal Plot
Visualizing convergence curves:

```python
fig, ax = plt.subplots(figsize=(6, 3.5), dpi=100)
x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x), color="#F37626", lw=2.5, label="Sine Wave")
ax.plot(x, np.cos(x), color="#2B7A78", lw=2.0, ls="--", label="Cosine Wave")
ax.set_title("Model Performance & Convergence")
ax.legend()
plt.show()
```

## 4. Summary & Observations
- The sine response reaches peak amplitude predictably.
- Organic users growth rate is **+52.4%** over the observation window.
- Ready for production deployment.
