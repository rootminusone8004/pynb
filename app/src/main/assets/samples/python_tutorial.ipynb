## 1. List Comprehensions & Lambdas

```python
squares = [x**2 for x in range(10) if x % 2 == 0]
double = lambda n: n * 2
print("Even squares:", squares)
print("Doubled:", [double(x) for x in squares])
```

## 2. Object Oriented Programming: Classes & Methods

```python
class DataProcessor:
    """A sample class demonstrating OOP in Python."""
    def __init__(self, name: str):
        self.name = name
        self.items = []
    
    def add_item(self, item):
        self.items.append(item)
        return self
        
    def summary(self):
        return f"{self.name}: {len(self.items)} items processed"

p = DataProcessor("Analytics Pipeline")
p.add_item("A").add_item("B").add_item("C")
print(p.summary())
```
